### Comprehensive Pandas Data Analysis



#### 1. Data Inspection and Overview

```python
print('DataFrame Info:')
display(df.info())

print('\nDataFrame Description (Numerical Columns):')
display(df.describe())

print('\nNumber of unique values per column:')
display(df.nunique())
```

#### 2. Column Selection

```python
print('Selecting a single column:')
display(df['customer_name'].head())

print('\nSelecting multiple columns:')
display(df[['customer_name', 'city', 'Orders.total_amount']].head())
```

#### 3. Row Selection (Filtering)

```python
print('Filtering rows based on a single condition (Orders from Bangalore):')
bangalore_orders = df[df['city'] == 'Bangalore']
display(bangalore_orders.head())

print('\nFiltering rows based on multiple conditions (High value orders in Bangalore):')
high_value_bangalore_orders = df[(df['city'] == 'Bangalore') & (df['Orders.total_amount'] > 5000)]
display(high_value_bangalore_orders.head())
```

#### 4. Grouping and Aggregation

```python
print('Total orders by payment method:')
payment_method_counts = df.groupby('Orders.payment_method')['Orders.order_id'].count().reset_index()
payment_method_counts.rename(columns={'Orders.order_id': 'total_orders'}, inplace=True)
display(payment_method_counts)

print('\nAverage total amount and sum of quantity ordered by Product Category:')
category_summary = df.groupby('Products.Category').agg(
    avg_total_amount=('Orders.total_amount', 'mean'),
    total_quantity_ordered=('Orders.quantity_ordered', 'sum')
).reset_index()
display(category_summary)
```

#### 5. Sorting Data

```python
print('Sorting by total order amount (descending):')
sorted_by_amount = df.sort_values(by='Orders.total_amount', ascending=False)
display(sorted_by_amount.head())

print('\nSorting by multiple columns (city then total amount):')
sorted_by_city_amount = df.sort_values(by=['city', 'Orders.total_amount'], ascending=[True, False])
display(sorted_by_city_amount.head())
```

#### 6. Creating New Columns (Data Transformation)

```python
print('Creating a new column `price_per_gm`:')
df['price_per_gm'] = df['Products.discountedSellingPrice'] / df['Products.weightInGms']
display(df[['Products.name', 'Products.discountedSellingPrice', 'Products.weightInGms', 'price_per_gm']].head())
```

#### 7. Time-based Analysis

```python
print('Extracting month and day of week from order date:')
df['order_month'] = df['Orders.order_date'].dt.month_name()
df['order_day_of_week'] = df['Orders.order_date'].dt.day_name()
display(df[['Orders.order_date', 'order_month', 'order_day_of_week']].head())

print('\nOrders per month:')
orders_per_month = df['order_month'].value_counts().reset_index()
orders_per_month.columns = ['Month', 'Number of Orders']
display(orders_per_month)
```

#### 8. Pivot Tables

```python
print('Pivot table showing average total amount by city and address type:')
pivot_table = df.pivot_table(values='Orders.total_amount', index='city', columns='address_type', aggfunc='mean')
display(pivot_table)
```

### 9. Visualization: Distribution of Orders.total_amount (Boxplot)

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(y=df['Orders.total_amount'])
plt.title('Distribution of Total Order Amount')
plt.ylabel('Total Order Amount')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 10. Visualization: Distribution of Orders.total_amount (Histogram)

```python
plt.figure(figsize=(12, 6))
sns.histplot(df['Orders.total_amount'], bins=50, kde=True)
plt.title('Distribution of Total Order Amount')
plt.xlabel('Total Order Amount')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 11. Visualization: Total Sales by Product Category (Bar Plot)

```python
# Ensure total_sales_by_category DataFrame is available from previous steps
# If not, recreate it:
# total_sales_by_category = df.groupby('Products.Category')['Orders.total_amount'].sum().reset_index()

plt.figure(figsize=(14, 7))
sns.barplot(x='Orders.total_amount', y='Products.Category', data=total_sales_by_category.sort_values(by='Orders.total_amount', ascending=False), palette='viridis')
plt.title('Total Sales by Product Category')
plt.xlabel('Total Sales Amount')
plt.ylabel('Product Category')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 12. Visualization: Discounted Selling Price vs. Total Order Amount (Scatter Plot)

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Products.discountedSellingPrice', y='Orders.total_amount', data=df, alpha=0.6)
plt.title('Discounted Selling Price vs. Total Order Amount')
plt.xlabel('Products Discounted Selling Price')
plt.ylabel('Total Order Amount')
plt.grid(linestyle='--', alpha=0.5)
plt.show()
```

### 13. Visualization: Sales by City (Column Chart / Bar Plot)

```python
sales_by_city = df.groupby('city')['Orders.total_amount'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='city', y='Orders.total_amount', data=sales_by_city.sort_values(by='Orders.total_amount', ascending=False), palette='viridis')
plt.title('Total Sales Amount by City')
plt.xlabel('City')
plt.ylabel('Total Sales Amount')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 14. Visualization: Monthly Sales Trend (Line Chart)

```python
monthly_sales = df.groupby('order_month')['Orders.total_amount'].sum().reindex(
    ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
).reset_index()
monthly_sales.columns = ['Month', 'Total Sales']

plt.figure(figsize=(12, 6))
sns.lineplot(x='Month', y='Total Sales', data=monthly_sales, marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 15. Visualization: Correlation Heatmap

```python
numerical_cols = ['Orders.total_amount', 'Orders.quantity_ordered', 'Products.mrp', 'Products.discountedSellingPrice', 'Products.weightInGms', 'Inventory.availableQuantity']
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

### 16. Visualization: Count of Orders by Address Type (Count Plot)

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='address_type', data=df, palette='pastel')
plt.title('Count of Orders by Address Type')
plt.xlabel('Address Type')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 18. Visualization: Kernel Density Estimate (KDE) Plot for Products.weightInGms

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(df['Products.weightInGms'], fill=True, color='purple')
plt.title('Kernel Density Estimate of Product Weight in Grams')
plt.xlabel('Product Weight in Gms')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
